# Python Project: Student Management System

## Project Description

Build a simple Student Management System using Python.

The program will help users add students, view student information, search for students, update grades, and calculate averages.

**Requirements:** Use Python basics only. No external libraries or databases are required.

---

```python
All_Students = []

def add_student():
    print("\n---------Add student---------")
    Student_Name = input("Enter Name of student: ")
    Student_ID = input("Enter ID of student: ")

    try:
        Student_Age = int(input("Enter Age of student: "))
    except ValueError:
        print("\ninvalid!!!\n")
        return

    try:
        Grades1 = float(input("Enter grade for subject 1: "))
        Grades2 = float(input("Enter grade for subject 2: "))
        Grades3 = float(input("Enter grade for subject 3: "))

        if 0 <= Grades1 <= 100 and 0 <= Grades2 <= 100 and 0 <= Grades3 <= 100:
            Grades = [Grades1, Grades2, Grades3]
            Student_data = {
                "Name": Student_Name,
                "ID": Student_ID,
                "Age": Student_Age,
                "Grades": Grades
            }
            All_Students.append(Student_data)
            print("\n--student Added successfully--\n")
        else:
            print("\nGrades must be between 0 and 100!!\n")
    except ValueError:
        print("\ninvalid!!!\n")
        return

def search_student():
    print("\n---------Search---------")
    target_id = input("Enter ID to search: ")
    found = False
    for student in All_Students:
        if student["ID"] == target_id:
            print(f"\nstudent is found : {student['Name']}\n")
            found = True
            break
    if not found:
        print("\nStudent not found\n")

def update_grades():
    print("\n---------Update Grades---------")
    target_id = input("Enter ID to Update grades: ")
    found = False
    for student in All_Students:
        if student["ID"] == target_id:
            try:
                g1 = float(input("Enter New grade for subject 1: "))
                g2 = float(input("Enter New grade for subject 2: "))
                g3 = float(input("Enter New grade for subject 3: "))
                if 0 <= g1 <= 100 and 0 <= g2 <= 100 and 0 <= g3 <= 100:
                    student["Grades"] = [g1, g2, g3]
                    print("\nGrades update successfully\n")
                else:
                    print("\nGrades must be between 0 and 100!\n")
            except ValueError:
                print("\ninvalid!\n")
            found = True
            break
    if not found:
        print("\nStudent not found\n")

def Calculate_avg():
    print("\n---------Calculate Average---------")
    target_id = input("Enter student id: ")
    found = False
    for student in All_Students:
        if student["ID"] == target_id:
            Average = sum(student["Grades"]) / len(student["Grades"])
            print(f"\naverage : {Average}")
            if Average >= 90:
                print("Grade: Excellent\n")
            elif Average >= 80:
                print("Grade: Very Good\n")
            elif Average >= 70:
                print("Grade: Good\n")
            elif Average >= 50:
                print("Grade: Pass\n")
            else:
                print("Grade: Fail\n")
            found = True
            break
    if not found:
        print("\nStudent not found\n")

def top_student():
    print("\n---------Top student---------")
    if not All_Students:
        print("\nNo students in the system yet!\n")
        return
    max_avg = 0
    top_studentname = ""
    for student in All_Students:
        Average = sum(student["Grades"]) / len(student["Grades"])
        if Average > max_avg:
            max_avg = Average
            top_studentname = student["Name"]
    print(f"\nthe top student is: {top_studentname}\n")

def Delete_student():
    print("\n---------Delete student---------")
    target_id = input("Enter student id: ")
    found = False
    for student in All_Students:
        if student["ID"] == target_id:
            deleted_name = student["Name"]
            All_Students.remove(student)
            print(f"\nstudent {deleted_name} deleted successfully\n")
            found = True
            break
    if not found:
        print("\nstudent not found\n")

def show_statistics():
    print("\n--------Show statistics----------")
    total = len(All_Students)
    print(f"Total number of students: {total}")

    if total == 0:
        print("\nNo students in the system yet!\n")
        return

    highest_avg = 0
    highest_name = ""
    lowest_avg = 100
    lowest_name = ""
    sum_averages = 0
    passing_count = 0
    failing_count = 0

    for student in All_Students:
        average = sum(student["Grades"]) / len(student["Grades"])
        sum_averages += average
        if average > highest_avg:
            highest_avg = average
            highest_name = student["Name"]
        if average < lowest_avg:
            lowest_avg = average
            lowest_name = student["Name"]

        if average >= 50:
            passing_count += 1
        else:
            failing_count += 1

    class_avg = sum_averages / total
    print(f"Highest average: {highest_avg} (Student: {highest_name})")
    print(f"Lowest average: {lowest_avg} (Student: {lowest_name})")
    print(f"Class average: {class_avg}")
    print(f"Number of passing students: {passing_count}")
    print(f"Number of failing students: {failing_count}\n")

while True:
    print("\n********************************")
    print("1. Add a New Student")
    print("2. Show All Students")
    print("3. Search for a Student")
    print("4. Update Student Grades")
    print("5. Calculate Student Average")
    print("6. Show the Top Student")
    print("7. Delete a Student")
    print("8. Show Statistics")
    print("9. Exit")

    choice = input("Choose an option: ")

    if choice == "9":
        print("\nExiting the system. Goodbye!\n")
        break
    elif choice == "1":
        add_student()
    elif choice == "2":
        print("\n---------All student---------")
        print(All_Students)
        print("-----------------------------\n")
    elif choice == "3":
        search_student()
    elif choice == "4":
        update_grades()
    elif choice == "5":
        Calculate_avg()
    elif choice == "6":
        top_student()
    elif choice == "7":
        Delete_student()
    elif choice == "8":
        show_statistics()
    else:
        print("\ninvalid\n")
```
